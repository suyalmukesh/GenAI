```python
# Returning multiple values 

def squared(x,y,z):

    return x*x , y*y , z*z 

```

```python
a , b , c = squared(2,3,4)

print(a)
print(b)
print(c)
```

```python
t = squared(2,3,4)

t
```

```python
xx = t[0]
yy = t[1]
zz = t[2]

print(xx,yy,zz)
```

```python
def func3(mylist):
    print("Inside function CALLED")
    print("Argument received : ",mylist)
    mylist.append(2)
    mylist.extend([5,1])
    print("Now my list becomes as : ",mylist)
    mylist.remove(5)
    print("List now becomes as ",mylist)
    return 

# __main__

list1 = [1]
print("[Line 14] My list before calling any fuinction ",list1)
func3(list1)
print("[Line 16 ] My list after function is called ",list1)

```

```python
def func4(mylist):
    print("Inside CALLED function")
    print("List received as : ",mylist)
    new = [3,5]
    mylist = new
    mylist.append(6)
    print("List within CALLED function , after changes ",mylist)
    return 

#__main__

List1 = [1,4]
print("[Line 13 ] List before the function call : ",List1)
func4(List1)
print("[Line 15 ] List after the function call : ",List1)
```
